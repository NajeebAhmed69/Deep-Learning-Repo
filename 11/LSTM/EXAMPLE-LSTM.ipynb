```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
```

```python
inputs = keras.Input(shape=(10, 3))
lstm_output, final_hidden, final_cell = layers.LSTM(
    units=32,
    return_state=True
)(inputs)
```

```python
prediction = layers.Dense(
    1,
    activation="linear"
)(lstm_output)
```

```python
prediction_model = keras.Model(
    inputs=inputs,
    outputs=prediction
)
```

```python
state_model = keras.Model(
    inputs=inputs,
    outputs=[
        lstm_output,
        final_hidden,
        final_cell
    ]
)
```

```python
prediction_model.compile(
    optimizer="adam",
    loss="mse"
)
```

```python
prediction_model.summary()
```

```python
state_model.compile(
    optimizer="adam",
    loss="mse"
)
```

```python
state_model.summary()
```

```python
import numpy as np

X = np.random.random((5, 10, 3))

predictions = prediction_model.predict(X)

print("Predictions:")
print(predictions)

states = state_model.predict(X)

print("LSTM Output Shape:", states[0].shape)
print("Hidden State Shape:", states[1].shape)
print("Cell State Shape:", states[2].shape)
```
