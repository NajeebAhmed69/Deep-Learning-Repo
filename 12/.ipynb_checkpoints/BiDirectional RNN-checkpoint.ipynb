```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.datasets import imdb
from tensorflow.keras.utils import pad_sequences

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    classification_report,
    ConfusionMatrixDisplay
)
```

```python

keras.utils.set_random_seed(42)

VOCAB_SIZE = 10_000
MAX_LENGTH = 200
EMBEDDING_DIM = 64
RNN_UNITS = 32

BATCH_SIZE = 128
EPOCHS = 8

print("TensorFlow version:", tf.__version__)
print("Available GPUs:", tf.config.list_physical_devices("GPU"))
```

```python
(x_train_full, y_train_full), (x_test_raw, y_test) = imdb.load_data(
    num_words=VOCAB_SIZE
)

print("Original training reviews:", len(x_train_full))
print("Test reviews:", len(x_test_raw))

print("\nFirst review: first 20 token IDs")
print(x_train_full[0][:20])

print("\nFirst review label:", y_train_full[0])
print("Label meaning: 0 = Negative, 1 = Positive")
```

```python
x_train_raw, x_val_raw, y_train, y_val = train_test_split(
    x_train_full,
    y_train_full,
    test_size=0.20,
    random_state=42,
    stratify=y_train_full
)

print("Training reviews:", len(x_train_raw))
print("Validation reviews:", len(x_val_raw))
print("Test reviews:", len(x_test_raw))
```

```python
def prepare_sequences(sequences):
    return pad_sequences(
        sequences,
        maxlen=MAX_LENGTH,
        padding="post",
        truncating="post",
        value=0
    )

x_train = prepare_sequences(x_train_raw)
x_val = prepare_sequences(x_val_raw)
x_test = prepare_sequences(x_test_raw)

print("Training input shape:", x_train.shape)
print("Validation input shape:", x_val.shape)
print("Test input shape:", x_test.shape)
```

```python
model = keras.Sequential([
    keras.Input(
        shape=(MAX_LENGTH,),
        dtype="int32",
        name="review_tokens"
    ),

    layers.Embedding(
        input_dim=VOCAB_SIZE,
        output_dim=EMBEDDING_DIM,
        mask_zero=True,
        name="word_embedding"
    ),

    layers.Bidirectional(
        layers.SimpleRNN(RNN_UNITS),
        merge_mode="concat",
        name="bidirectional_rnn"
    ),

    layers.Dropout(0.3),

    layers.Dense(
        32,
        activation="relu",
        name="dense_features"
    ),

    layers.Dense(
        1,
        activation="sigmoid",
        name="sentiment_probability"
    )
])

model.summary()
```

```python
model.compile(
    optimizer=keras.optimizers.Adam(
        learning_rate=0.001,
        global_clipnorm=1.0
    ),
    loss="binary_crossentropy",
    metrics=[
        keras.metrics.BinaryAccuracy(name="accuracy"),
        keras.metrics.AUC(name="auc")
    ]
)
```

```python
early_stopping = keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=2,
    restore_best_weights=True
)

history = model.fit(
    x_train,
    y_train,
    validation_data=(x_val, y_val),
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    callbacks=[early_stopping],
    verbose=1
)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

epochs_run = range(1, len(history.history["loss"]) + 1)

axes[0].plot(
    epochs_run,
    history.history["accuracy"],
    marker="o",
    label="Training accuracy"
)
axes[0].plot(
    epochs_run,
    history.history["val_accuracy"],
    marker="o",
    label="Validation accuracy"
)
axes[0].set_title("Accuracy")
axes[0].set_xlabel("Epoch")
axes[0].set_ylabel("Accuracy")
axes[0].legend()
axes[0].grid(alpha=0.3)

axes[1].plot(
    epochs_run,
    history.history["loss"],
    marker="o",
    label="Training loss"
)
axes[1].plot(
    epochs_run,
    history.history["val_loss"],
    marker="o",
    label="Validation loss"
)
axes[1].set_title("Loss")
axes[1].set_xlabel("Epoch")
axes[1].set_ylabel("Binary cross-entropy")
axes[1].legend()
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
test_results = model.evaluate(
    x_test,
    y_test,
    batch_size=BATCH_SIZE,
    return_dict=True,
    verbose=1
)

for metric_name, metric_value in test_results.items():
    print(f"{metric_name}: {metric_value:.4f}")
```

```python
test_probabilities = model.predict(
    x_test,
    batch_size=BATCH_SIZE,
    verbose=1
).ravel()

test_predictions = (test_probabilities >= 0.5).astype("int32")

print(
    classification_report(
        y_test,
        test_predictions,
        labels=[0, 1],
        target_names=["Negative", "Positive"],
        digits=4,
        zero_division=0
    )
)

ConfusionMatrixDisplay.from_predictions(
    y_test,
    test_predictions,
    labels=[0, 1],
    display_labels=["Negative", "Positive"],
    cmap="Blues",
    values_format="d"
)

plt.title("Bidirectional RNN — IMDB Test Set")
plt.tight_layout()
plt.show()
```

```python
word_index = imdb.get_word_index()

index_to_word = {
    index + 3: word
    for word, index in word_index.items()
}

index_to_word.update({
    0: "[PAD]",
    1: "[START]",
    2: "[OOV]",
    3: "[UNUSED]"
})

def decode_review(sequence):
    return " ".join(
        index_to_word.get(int(token), "[OOV]")
        for token in sequence
        if token != 0
    )

for i in range(3):
    probability = float(test_probabilities[i])

    predicted_label = "Positive" if probability >= 0.5 else "Negative"
    actual_label = "Positive" if y_test[i] == 1 else "Negative"

    print(f"\nReview {i + 1}")
    print("-" * 70)
    print(decode_review(x_test[i]))
    print("\nActual sentiment:", actual_label)
    print("Predicted sentiment:", predicted_label)
    print(f"Positive probability: {probability:.4f}")
```
